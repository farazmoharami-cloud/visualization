## Course introduction notes

### Why data visualization matters

- Data visualization combines evidence, design, perception, and storytelling.
- Historical examples include John Snow's cholera map, Florence
  Nightingale's advocacy graphics, and Charles Minard's visualization of Napoleon's
  Russian campaign.
- The Challenger example contrasts Edward Tufte's interpretation with William
  Robison's critique. The central lesson is that visualization choices influence what
  an audience notices and what conclusions seem reasonable.

### Three qualities used to assess a visualization

- **Aesthetic:** Is the visualization visually coherent and engaging?
- **Substantive:** Does it communicate meaningful, appropriately supported content?
- **Perceptual:** Can viewers accurately perceive and compare the encoded values?

Context also matters: the audience, purpose, medium, and structure of the data should
shape the design.


## Visual critique activity

- **3D life-expectancy books:** Eye-catching, but the 3D perspective distorts the
  values. Labels and provenance are unclear, so the substantive and perceptual
  qualities are weak.
- **"Monstrous Costs" dragon:** Memorable and expressive, but the decorative form
  makes precise comparisons difficult.
- **Same data with different aspect ratios:** The apparent steepness changes even
  though the values do not. Aspect ratio can therefore change the perceived story.
- **Presidential-run pie chart:** Visually busy, and the percentages are not a clear
  part-to-whole relationship. A bar chart would support comparison better.
- **Wind map:** Strong aesthetically and effective for seeing spatial patterns.
  Interpretation still depends on a clear legend, source, time frame, and interactive
  context.


## Matplotlib imports


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

## Sample data and first scatterplot


```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y)
```

## Bar and line plots


```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x, y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
```

## Histogram activity


```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

## Titles and axis labels


```python
ax.set_title("Total growth over time")
ax.set_ylabel("Total growth")
ax.set_xlabel("Years since start")
fig.tight_layout()
```

## Font dictionaries and styled labels


```python
font1 = {"family": "sans-serif", "color": "blue", "size": 20}
font2 = {"family": "monospace", "color": "green", "size": 14}

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_title("Total growth over time", fontdict=font1)
ax.set_ylabel("Total growth", fontdict=font2)
ax.set_xlabel("Years since start", fontdict=font2)
fig.tight_layout()
```

## Label formatting


```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_ylabel("Total growth", fontdict=font2)
ax.set_xlabel("Years since start", fontdict=font2)
```

## Customizing points and lines


```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(
    x,
    y,
    marker="*",       # NEW ADDITION
    color="indigo",   # NEW ADDITION
)

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(
    x,
    y,
    marker="*",
    color="indigo",
    linestyle="--",   # NEW ADDITION
    linewidth=2,      # NEW ADDITION
)

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

# color is the new addition
ax.plot(x, y, marker="*", color="#7425b9", linestyle="--", linewidth=2)

fig.show()
```

```python
ax.plot(
    x,
    y,
    marker="*",
    markersize=12,                # NEW ADDITION
    color="#7425b9",
    linestyle="--",
    linewidth=2,
    markeredgecolor="#fa9359",    # NEW ADDITION
    markerfacecolor="#000000",    # NEW ADDITION
)
```

## Grid-line activity


```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(
    x,
    y,
    marker="*",
    markersize=12,
    color="#7425b9",
    linestyle="--",
    linewidth=2,
    markeredgecolor="#fa9359",
    markerfacecolor="#000000",
)
ax.grid(axis="y")
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(
    x,
    y,
    marker="*",
    markersize=12,
    color="#7425b9",
    linestyle="--",
    linewidth=2,
    markeredgecolor="#fa9359",
    markerfacecolor="#000000",
)
ax.grid(axis="y")
ax.grid(axis="y", color="blue", linewidth=2, linestyle="-.")
fig.show()
```
