## Activity: Is a visualization neutral?

- The Periscopic gun-death visualization foregrounds years of potential life lost.
  Its animation and long trajectories create an emotional emphasis while still being
  based on factual records.
- The Washington Post active-shooter graphic uses restrained color, familiar bars,
  and a compact layout. It may appear more neutral, but its categories, ordering, and
  framing are still rhetorical choices.
- The Republican and Democratic versions of the same jobs report emphasize different
  comparisons: one highlights being below a target while the other highlights the
  decline from a prior peak. Both may use accurate numbers while telling different
  stories.

**Conclusion:** A visualization is not neutral. It can be factually responsible while
still making rhetorical choices about inclusion, comparison, scale, color, and focus.


## Choosing a chart

Consider:

- purpose, audience, and medium;
- the type and attributes of the data;
- whether the task is comparison, distribution, relationship, composition, change
  over time, or spatial pattern;
- how accurately viewers can decode the chosen marks and channels.

## Gestalt principles

- **Proximity:** nearby marks appear related.
- **Similarity:** marks with similar color or shape appear grouped.
- **Continuity:** viewers follow smooth paths and alignments.
- **Closure:** viewers mentally complete incomplete forms.
- **Enclosure:** a boundary groups the marks inside it.
- **Connection:** connected marks appear strongly related.

## Cognitive load and trust

- Prefer familiar forms when an audience needs a quick explanation.
- Position on a shared scale is generally easier to compare than angle, area, volume,
  or 3D perspective.
- Explanatory graphics should be concise; exploratory graphics may support more
  detail and interaction.
- Put essential information in the graphic instead of relying on memory.
- Clear provenance, source notes, and methods increase trust.
