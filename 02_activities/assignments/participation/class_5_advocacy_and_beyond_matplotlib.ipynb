# Class 5: Data Visualization as Advocacy and Beyond Matplotlib

## Advocacy notes and activity responses

Visualization can persuade through:

- **Rational appeal:** evidence, measurement, comparison, and logic.
- **Moral appeal:** values, justice, responsibility, and human dignity.
- **Emotional appeal:** empathy, urgency, fear, grief, or hope.

### Brooks diagram activity

- Rational: the spatial plan, counts, and density document how people were confined.
- Moral: people are shown as being treated as cargo, making the injustice visible.
- Emotional: the repeated cramped bodies communicate horror and loss of humanity.

### Presentation and representation

Presentation concerns how information is arranged and encoded. Representation asks
whose experiences are included, how categories are defined, and what the data leaves
out. Binary gender categories, for example, can exclude people and hide variation.
Unconventional data and lived experience may reveal things that institutional data
miss.

### Credit the underwater labour

Visualization work may depend on community organizers, palette designers, research
assistants, technical writers, IT staff, caregivers, and many others. Contributions
should be acknowledged instead of crediting only the final visible author.


## Imports and Seaborn data


```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

tips = sns.load_dataset("tips")
print(tips)
```

## Seaborn line plot, style, and labels


```python
sns.lineplot(
    data=tips,       # choose our dataset
    x="total_bill",  # define our x variable
    y="tip",         # define our y variable
)
```

```python
sns.set_style("whitegrid")

tipgraph = sns.lineplot(
    data=tips,
    x="total_bill",
    y="tip",
)

tipgraph.set(
    title="Tips vs. Total Bill",
    xlabel="Total Bill ($)",
    ylabel="Tip Amount ($)",
)
```

```python
fig = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(
    data=tips,
    x="total_bill",
    y="tip",
    color="hotpink",
    linestyle="--",
    linewidth=3,
    marker="o",
    markerfacecolor="indigo",
)
```

## Seaborn scatterplot and pairplot


```python
tipgraph = sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    style="time",
    hue="day",
    palette=["purple", "hotpink", "deepskyblue", "yellowgreen"],
)

tipgraph.set(
    title="Tips vs. Total Bill",
    xlabel="Total Bill ($)",
    ylabel="Tip Amount ($)",
)
```

```python
sns.pairplot(
    data=tips,
    hue="day",
)
```

## Faceted relational plot and activity comments


```python
daysplot = sns.relplot(
    data=tips,          # dataset used in the plot
    x="total_bill",     # variable on the x-axis
    y="tip",            # variable on the y-axis
    hue="sex",          # use colour to represent sex
    col="day",          # make a subplot for each day
    kind="scatter",     # make scatterplots
    col_wrap=2,         # wrap the subplots into two columns
)
```

## Plotly bar chart and HTML export


```python
import plotly.graph_objects as go  # "go" means graph objects

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])

graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))

graph.update_layout(
    title="Pirate Scores",
    xaxis_title="Pirates",
    yaxis_title="Score",
)

graph.show()
```

```python
graph.write_html("pirategraph.html")
```

## Plotly scatter customization


```python
graph = go.Figure()
graph.add_trace(
    go.Scatter(
        x=x1,
        y=y1,
        mode="markers",  # points for a scatterplot
        marker=dict(
            size=15,                            # point size
            color="hotpink",                    # point colour
            opacity=1,                          # point transparency
            line=dict(width=5, color="purple"), # point outline
        ),
    )
)

graph.update_layout(
    title="Interactive Pirate Plot",
    xaxis_title="Pirates",
    yaxis_title="Scores",
    width=500,
    height=500,
)
```

## Word cloud


```python
import numpy as np
from wordcloud import WordCloud

df = pd.read_csv(
    "https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
    on_bad_lines="skip",
)

text = " ".join(each for each in df.quote)

wordcloud = WordCloud(
    background_color="white",
    colormap="inferno",
).generate(text)

fig, ax = plt.subplots(figsize=(7, 3))

# Convert the WordCloud image manually to a NumPy array
ax.imshow(np.array(wordcloud.to_image()), interpolation="bilinear")

ax.axis("off")
plt.show()
```

## Venn diagram


```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2_unweighted(
    [A, B],
    set_labels=("Fruits", "Pies"),
    set_colors=("blue", "red"),
    alpha=0.5,
)
plt.show()
```
